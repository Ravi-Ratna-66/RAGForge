```python
X = np.array([500,1000,1500,2000])
Y = np.array([20,40,60,80])
w = 0.04
b = 0
predict = w*X+b
print(predict)
error = Y - predict
mse = np.mean(error**2)
print(mse)
```

```python
gradient_w = -2 * np.mean(X * error)
gradient_b = -2 * np.mean(error)

print("gradient w:", gradient_w)
print("gradient b:", gradient_b)
```

```python
learning_rate = 0.00001

w = w - learning_rate * gradient_w
b = b - learning_rate * gradient_b
print("new w:", w)
print("new b:", b)
```

```python
for i in range(1000):

    predictions = w * X + b

    error = Y - predictions

    gradient_w = -2 * np.mean(X * error)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b
print("w:", w)
print("b:", b)

predictions = w * X + b
print("Predictions:", predictions)

mse = np.mean((Y - predictions) ** 2)
print("MSE:", mse)
```

```python
X = np.array([0.5, 1, 1.5, 2, 2.5, 3])
Y = np.array([23, 38, 61, 76, 105, 117])

w = 0
b = 0

learning_rate = 0.01

for i in range(1000):
    predictions = w * X + b

    error = Y - predictions

    gradient_w = -2 * np.mean(X * error)
    gradient_b = -2 * np.mean(error)

    w = w - learning_rate * gradient_w
    b = b - learning_rate * gradient_b

predictions = w * X + b
mse = np.mean((Y - predictions) ** 2)

print("w:", w)
print("b:", b)
print("Predictions:", predictions)
print("MSE:", mse)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, Y, label="Actual data")

plt.plot(X, predictions, label="Model")

plt.xlabel("House size (thousand sq ft)")
plt.ylabel("Price (lakh)")
plt.title("Linear Regression from Scratch")

plt.legend()
plt.show()
```
