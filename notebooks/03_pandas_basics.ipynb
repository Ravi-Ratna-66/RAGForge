```python
import pandas as pd
data = {
    "Name": ["Ravi", "Alex", "Sam", "John"],
    "Age": [21, 24, 22, 25],
    "Salary": [50000, 65000, 55000, 72000],
    "Department": ["AI", "ML", "AI", "Data"]
}

df = pd.DataFrame(data)

print(df)
print(df.shape)
```

```python
df_missing = df.copy()
df_missing.loc[1, "Age"] = None
df_missing.loc[2, "Salary"] = None
print(df_missing)
```

```python
print(df_missing.isnull())
print(df_missing.isnull().sum())
```

```python
print("Mean Age:", df_missing["Age"].mean())
print("Mean Salary:", df_missing["Salary"].mean())
```

```python
df_missing["Age"] = df_missing["Age"].fillna(df_missing["Age"].mean())

df_missing["Salary"] = df_missing["Salary"].fillna(df_missing["Salary"].mean())
print(df_missing)
print(df_missing.isnull().sum())
```

```python
import pandas as pd
df = pd.read_csv("C:\RAGForge\data\employees.csv")
print(df)
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nMissing values:")
print(df.isnull().sum())

print("\nStatistics:")
print(df.describe())
```

```python
print(df.sort_values("Salary", ascending=False).head(3))
print(df["Department"].value_counts())

print(df.groupby("Department")["Salary"].mean())

print(df[df["Age"] > 25])
```

```python
df_messy = df.copy()

df_messy = pd.concat(
    [df_messy, df_messy.iloc[[2]]],
    ignore_index=True
)

print("Duplicates:")
print(df_messy.duplicated())

print("\nNumber of duplicates:", df_messy.duplicated().sum())
df_messy = df_messy.drop_duplicates()

print(df_messy.shape)
```

```python
df_messy.loc[5, "Salary"] = None
print(df_messy.isnull().sum())
print("Mean:", df_messy["Salary"].mean())
print("Median:", df_messy["Salary"].median())
```

```python
df_messy["Salary"] = df_messy["Salary"].fillna(df_messy["Salary"].median())
print(df_messy)
print("\nMissing values:")
print(df_messy.isnull().sum())
```

```python
df_encoded = pd.get_dummies(
    df_messy,
    columns=["Department"],
    dtype=int
)
df_encoded = df_encoded.drop(columns=["Name"])
print(df_encoded)
```
