```python
x = torch.tensor(2.0, requires_grad=True)

print(x)
print(x.requires_grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x ** 2

y.backward()

print("x:", x)
print("y:", y)
print("Gradient:", x.grad)
```

```python
w = torch.tensor(2.0, requires_grad=True)

loss = (w - 5) ** 2

loss.backward()

print("w:", w)
print("Loss:", loss)
print("Gradient:", w.grad)
```

```python
w = torch.tensor(2.0, requires_grad=True)

learning_rate = 0.1

for step in range(10):
    loss = (w - 5) ** 2

    loss.backward()

    with torch.no_grad():
        w -= learning_rate * w.grad

    w.grad.zero_()

    print(
        "Step:", step + 1,
        "w:", w.item(),
        "Loss:", loss.item()
    )
```
